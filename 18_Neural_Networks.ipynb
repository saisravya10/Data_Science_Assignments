```python
features = df.drop(columns= ['Y'])
target = df['Y']
```

```python
#encoding target i.e, converting into numbers (1D)
le = LabelEncoder()
y = le.fit_transform(target)
# One-hot encode the target variable to match the model's output layer units
y = tf.keras.utils.to_categorical(y)
```

```python
#normalise features
std_sca = StandardScaler()
features = std_sca.fit_transform(features)
```

# **`2. Model Implementation`**



**●	Construct a basic ANN model using your chosen high-level neural network library. Ensure your model includes at least one hidden layer.**


```python
epochs = 10
n_classes = 2
batch_size = 100
```

```python
#Model buliding

ann = tf.keras.models.Sequential()
#input layer
ann.add(tf.keras.layers.Dense(units=42, activation='relu', input_shape=[60]))
#hidden layer
ann.add(tf.keras.layers.Dense(units=30, activation='relu'))
#output layer
ann.add(tf.keras.layers.Dense(units=n_classes, activation='sigmoid'))

ann.summary()
```

```python
#Model compile

ann.compile(optimizer=tf.optimizers.Adam(learning_rate=0.001), loss=tf.keras.losses.binary_crossentropy, metrics=['accuracy'])
```

**●	Divide the dataset into training and test sets.**


```python
#splitting data into training and testing
x_train, x_test, y_train, y_test = train_test_split(features, y, train_size=0.8, random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

**●	Train your model on the training set and then use it to make predictions on the test set.**

```python
#Model training

history = ann.fit(x_train, y_train, epochs=epochs, batch_size=batch_size, validation_data=(x_test, y_test))
```

```python
#Model validation

ann.evaluate(x_test, y_test)
```

```python
#Prediction

y_pred = ann.predict(x_test)
y_pred
```

```python
#cross checking whether we getting same values for predicted and original y_test
print(np.argmax(y_pred[0]))
print(np.argmax(y_test[0]))
```

```python
print(np.argmax(y_pred[40]))
print(np.argmax(y_test[40]))
```

Explanation:
---

In this task, an Artificial Neural Network (ANN) model was implemented to classify the dataset into two classes. The dataset was divided into training and testing sets using an 80:20 split.

The ANN model was built using the Sequential. It consists of an input layer with 42 neurons and ReLU activation, one hidden layer with 30 neurons and ReLU activation, and an output layer with 2 neurons using sigmoid activation.

The model was compiled using the Adam optimizer with a learning rate of 0.001 and binary cross-entropy as the loss function. Accuracy was used as the evaluation metric. The model was trained for a fixed number of epochs and validated using the test dataset. Finally, predictions were generated and compared with the actual labels to verify model performance

**●	Adopt a structured approach like grid search or random search for hyperparameter tuning, documenting your methodology thoroughly.**

```python
def create_model(neurons=32, activation='relu', learning_rate=0.001):
    model = tf.keras.models.Sequential()
    model.add(tf.keras.layers.Dense(neurons, activation=activation, input_shape=(x_train.shape[1],)))
    model.add(tf.keras.layers.Dense(n_classes, activation='sigmoid'))

    optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)
    model.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])
    return model
```

```python
ann_model = KerasClassifier(build_fn=create_model, epochs=10, batch_size=32, verbose=0)
```

```python
param_grid = {
    'model__neurons': [16, 32],
    'model__activation': ['relu', 'tanh'],
    'model__learning_rate': [0.001, 0.01]
}
```

```python
# Grid Search
grid_search = GridSearchCV(estimator=ann_model, param_grid=param_grid, cv=3, scoring='accuracy')
```

```python
grid_search.fit(x_train, y_train)
```

```python
# Best parameters
print("Best Parameters:", grid_search.best_params_)
print("Best Accuracy:", grid_search.best_score_)
```

Explanation:
---

A structured hyperparameter tuning approach was adopted using Grid Search to improve the performance of the ANN model. The hyperparameters tuned include the number of neurons, activation function, learning rate, batch size, and number of epochs. GridSearchCV was used to systematically evaluate different combinations of these hyperparameters using cross-validation and select the best configuration based on accuracy.

⸻

From the grid search results, it was observed that the model performed best when using the ReLU activation function with a learning rate of 0.01 and 16 neurons in the hidden layer. Additionally, a batch size of 80 and 10 epochs provided better convergence and improved accuracy compared to other combinations. This confirms that appropriate selection of hyperparameters significantly enhances the model’s performance.
