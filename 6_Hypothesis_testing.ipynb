# **`ESTIMATION AND CONFIDENCE INTERVALS`**

Durability of print-heads(in millions of characters) - Sample Data


(1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29)

```python
#loading the sample data
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

**a. Build 99% Confidence Interval Using Sample Standard Deviation**

```python
#Given data
n = 15 #lenght of data
CI = 0.99 #CI = 99%
alpha = 0.01 #alpha = 100-CI = 1%
```

```python
#calucalte sample mean and standard deviation
x_bar = np.mean(data) #sample mean
s = np.std(data, ddof=1) #sample standarad deviation
print(x_bar, s)
```

```python
#choose the distribution
#Here population standarad deviation is unknown and also n<30, so we will go with t-distribution and perform t-test

df = n-1 #degree of freedom
df
```

```python
#set confidence level & finding area
CI = 0.99
alpha = 0.01 #alpha/2 = 0.005
area = CI + alpha/2 #area for postive boundary
area
```

```python
#find t-criticial value
t_critical = stats.t.ppf(area, df)
t_critical
```

```python
#margin of error
margin_of_error_t_test = t_critical * (s/np.sqrt(n))
margin_of_error_t_test
```

```python
#confidence interval from t-test
confidence_interval_t_test = (x_bar - margin_of_error_t_test, x_bar + margin_of_error_t_test)
confidence_interval_t_test
```

**Explanation**

First, the sample data was loaded and the given details such as sample size, confidence level, and alpha value were identified. The sample mean and sample standard deviation were then calculated from the data. Since the population standard deviation is unknown and the sample size is less than 30, the t-distribution was used. For a 99% confidence interval, the area 0.995 was chosen to obtain the positive t-critical value. Using this t-critical value from above step, the margin of error was calculated. Finally, the confidence interval was obtained.

**b. Build 99% Confidence Interval Using Known Population Standard Deviation**

```python
#given data
pop_std_dev = 0.2 #sigma value
```

```python
#choose the distribution
#Here population standarad deviation is known, so we will go with z-distribution and perform z-test
#requried data
CI = 0.99
alpha = 0.01 #alpha/2 = 0.005
pop_std_dev = 0.2
area = 0.995
print(x_bar)
```

```python
#find z-critical value
z_critical = stats.norm.ppf(area)
z_critical
```

```python
#margin of error
margin_of_error_z_test = z_critical * (pop_std_dev/np.sqrt(n))
margin_of_error_z_test
```

```python
#confidence interval from z-test
confidence_interval_z_test = (x_bar - margin_of_error_z_test, x_bar + margin_of_error_z_test)
confidence_interval_z_test
```

**Explanation**

Using the above data, the sample mean(x_bar) was taken, and the population standard deviation was provided as 0.2. The confidence level was chosen as 99%, so the alpha value became 0.01. So the area 0.995 was used to find the positive z-critical value. Because the population standard deviation is known, the z-distribution was used instead of the t-distribution. With the z-critical value, the margin of error was calculated, and finally the confidence interval was obtained.

# **`HYPOTHESIS TESTING`**

**Given Data**

Weekly operatingcost model(W) = 1000+5X

Mean weekly cost of 25 sample restaurants = 3050

Number of units produced in a week(X) mean = 600

Number of units produced in a week(X) standard deviation = 25







```python
n = 25
x_bar = 3050
pop_mean = 4000 #X = 600, (W = 1000+5(600) = 4000)
pop_std_dev = 125 #(5 units i.e, 5*25 = 125)
alpha = 0.05
CI = 0.95
```

**1. State the Hypotheses statement:**

H0 = The mean weekly operating cost follows the given cost model (=4000)

Ha = The mean weekly operating cost is higher than the cost model suggests (>4000)

**2. Calculate the Test Statistic:**

```python
#using z-test
z_score = (x_bar - pop_mean)/(pop_std_dev/np.sqrt(n))
z_score
```

**3. Determine the Probability and compare:**

```python
z_critical = stats.norm.ppf(1-alpha)
z_critical
```

since z_score < z_critical, it doesn't fall in the rejection region

**4. Make a Decision:**

Since the test statistic doesn't fall in the rejection region, we fail to reject the null hypothesis(H0)

**5. Conclusion:**

There is no strong statistical evidence to support the restaurant owners' claim that the weekly operation costs are higher than what the model suggets.
