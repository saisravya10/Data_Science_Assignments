1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
#create
list = [2, 4, 6, 8, 10]
print(list)
```

```python
#append
list.append(12)
print(list)
```

```python
#extend
list.extend([14, 16, 18, 20])
print(list)
```

```python
#insert
list.insert(2, 5)
print(list)
```

```python
#remove
list.remove(12)
print(list)
```

```python
#pop
popped_value = list.pop(5)
print("popped_value: ", popped_value)
print(list)
```

```python
#index
index_of_value_18 = list.index(18)
print("index_of_value_18: ", index_of_value_18)
```

```python
#count
count_5 = list.count(5)
print("count_5:", count_5)
```

```python
#sort(ascending order)
list.sort()
print(list)
```

```python
#reverse
list.sort(reverse = True)
print(list)
```

```python
#clear
list.clear()
print(list)
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
info = {
    'Name': ['Chandu', 'Thanuja', 'Priya', 'Sirisha', 'Meghana',
             'Balu', 'Harish', 'Sai', 'Prabhas', 'Ian'],
    'Age': [25, 28, 22, 18, 27, 35, 29, 24, 26, 25],
    'City': ['Delhi', 'Mumbai', 'Hyderabad', 'Chennai', 'Pune',
             'Vizag', 'Kolkata', 'Bangalore', 'Pune', 'Delhi'],
    'Salary': [40000, 50000, None, 60000, 55000,
               70000, None, 48000, 52000, 40000]
}
```

```python
df = pd.DataFrame(info)
df
```

Use .info(), .describe()

```python
#info
df.info()
```

```python
#describe
df.describe()
```

Select 'Name' and 'City' columns

```python
df[['Name', 'City']]
```

Drop 'City' column

```python
df= df.drop(columns = ['City'])
df
```

Fill any missing values in 'Salary' column with the mean

```python
salary_mean = df['Salary'].mean()
df['Salary'].fillna(salary_mean, inplace = True)
df
```

Remove any duplicate rows

```python
df = df.drop_duplicates()
df
```
